# Titanic: Data Health, Feature Effects, Baseline

Before building and training models, it is often useful to understand:

- how clean is the data
- are there synthetic features that add value
- which features matter most
- what a simple model looks like
- how the model compares to a naive baseline

These checks help avoid spending time optimizing models on datasets that may have deeper issues.
The diagnostics shown below were generated automatically using a dataset analysis tool.


## Data Health

![Data Health](titanic_data_health.png)

The data health reports highlight issues such as:

- missing values
- low-variance columns
- potential target imbalance

This can help prioritize which columns may need cleaning, transformation, or removal before training models.


## Feature Effects
